# Clean — NHANES 2015-2016 CBC_I

**Vấn đề từ EDA**: chất lượng đã cao, chỉ có 1.049/9.165 dòng thiếu phần 5-dòng bạch cầu chi tiết.

**Quyết định**: xuất 2 phiên bản — bản đầy đủ (9.165 dòng, dùng khi chỉ cần CBC cơ bản) và bản chỉ giữ ca hoàn chỉnh 100% (8.116 dòng, dùng khi cần đủ cả 5-dòng bạch cầu).

**Output**: `Clean_Data/tabular/nhanes_cbc_full.csv` + `nhanes_cbc_complete_only.csv`

```python
import pandas as pd
from pathlib import Path

DATA = Path(r'D:\AI_08_V1\Data')
OUT = Path(r'D:\AI_08_V1\Clean_Data\tabular')

df = pd.read_csv(DATA / 'tabular' / 'nhanes_cbc_2015' / 'CBC_I_with_demographics.csv')
print('Shape gốc:', df.shape)
print('Trùng lặp:', df.duplicated().sum())

diff_cols = ['LBXLYPCT', 'LBXMOPCT', 'LBXNEPCT', 'LBXEOPCT', 'LBXBAPCT']
df.to_csv(OUT / 'nhanes_cbc_full.csv', index=False)
print(f'Đã lưu bản đầy đủ: {len(df)} dòng')

df_complete = df.dropna(subset=diff_cols).reset_index(drop=True)
df_complete.to_csv(OUT / 'nhanes_cbc_complete_only.csv', index=False)
print(f'Đã lưu bản hoàn chỉnh 100%: {len(df_complete)} dòng')
```
