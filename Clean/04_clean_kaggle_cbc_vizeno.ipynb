# Clean — Kaggle CBC (vizeno)

**Vấn đề từ EDA**: 77.5% dòng trùng lặp (323/417) — mức cao nhất kho.

**Quyết định**: `drop_duplicates()`. Cỡ mẫu sau khi làm sạch sẽ rất nhỏ (~94 dòng) — chỉ dùng bổ sung/validation, không dùng làm nguồn train chính.

**Output**: `Clean_Data/tabular/kaggle_cbc_vizeno_clean.csv`

```python
import pandas as pd
from pathlib import Path

DATA = Path(r'D:\AI_08_V1\Data')
OUT = Path(r'D:\AI_08_V1\Clean_Data\tabular')

df = pd.read_csv(DATA / 'tabular' / 'kaggle_cbc_vizeno' / 'blood_count_dataset.csv')
print('Shape gốc:', df.shape)

df_clean = df.drop_duplicates().reset_index(drop=True)
print(f'Sau loại trùng: {len(df_clean)} dòng ({len(df_clean)/len(df)*100:.1f}% giữ lại)')

df_clean.to_csv(OUT / 'kaggle_cbc_vizeno_clean.csv', index=False)
print(f'Đã lưu: {OUT / "kaggle_cbc_vizeno_clean.csv"} ({len(df_clean)} dòng)')
```
