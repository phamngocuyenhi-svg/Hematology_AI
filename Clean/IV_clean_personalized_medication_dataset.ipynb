**Input:** `Dataset/Data/personalized_medication_dataset.csv`

## Vấn đề phát hiện từ EDA (IV_personalized_medication_dataset.ipynb)

| # | Vấn đề | Mô tả | Mức độ ảnh hưởng |
|---|--------|-------|------------------|
| 1 | Missing values | 5 cột có giá trị thiếu: Chronic_Conditions (246), Drug_Allergies (324), Genetic_Disorders (339), Recommended_Medication (263), Dosage (195), Duration (257) | Trung bình - cần impute trước khi modeling |
| 2 | Duplicates | 0 hàng trùng lặp | Thấp |
| 3 | Biological outliers | Không có giá trị vô lý ngoài khoảng sinh học | Thấp |
| 4 | Imbalanced labels | 5 lớp cân bằng tốt (~180-220 mẫu/lớp) | Thấp |
1. **Strip columns:** Xóa khoảng trắng thừa ở tên cột.
2. **Strip text values:** Xóa khoảng trắng thừa ở các cột text.
3. **Missing values:** Impute các cột categorical/text với giá trị `"Unknown"`.
4. **Duplicates:** Không có, giữ nguyên.
5. **Outliers:** Không có, giữ nguyên.
6. **Export:** Lưu kết quả ra `Clean_Data/personalized_medication_dataset_clean.csv`.

---
  

```python
import pandas as pd
import numpy as np
import os

# ============================================
# CONFIGURATION
# ============================================
INPUT_PATH = "../Data/personalized_medication_dataset.csv"
OUTPUT_DIR = "../Clean_Data"
OUTPUT_FILE = os.path.join(OUTPUT_DIR, "personalized_medication_dataset_clean.csv")
```

```python
# Đọc dữ liệu
df = pd.read_csv(INPUT_PATH)
print(f"Original shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.head(5)
# Strip whitespace from column names
df.columns = df.columns.str.strip()
print('Columns after strip:', list(df.columns))
df.head(5)
```

```python
# Missing values
missing = df.isnull().sum()
missing_sum = missing.sum()
print('=== Missing Values ===')
if missing_sum == 0:
    print('Không có missing values.')
else:
    print(f'Tổng missing values: {missing_sum}')
    print(missing[missing > 0])

# Duplicates
duplicates = df.duplicated().sum()
print(f'\n=== Duplicates ===')
print(f'Số hàng trùng lặp: {duplicates}')
```

```python
# Missing values by column
# Các cột categorical/text có missing values
text_cols = ['Chronic_Conditions', 'Drug_Allergies', 'Genetic_Disorders', 'Recommended_Medication', 'Dosage', 'Duration']

# Impute missing values với 'Unknown'
for col in text_cols:
    if col in df.columns:
        missing_before = df[col].isnull().sum()
        df[col] = df[col].fillna('Unknown')
        missing_after = df[col].isnull().sum()
        print(f'{col}: {missing_before} missing -> {missing_after} missing (filled with "Unknown")')

# Strip whitespace from text values
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].str.strip()
```

```python
print("=== FINAL DATA SUMMARY ===")
print(f"Shape: {df.shape}")
print(f"\nData types:")
print(df.dtypes)
print(f"\n=== Missing Values ===")
print(f"Missing: {df.isnull().sum().sum()}")
print(f"\n=== Duplicates ===")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"\n=== Label Distribution ===")
print(df['Diagnosis'].value_counts())
```

```python
# Export cleaned data
os.makedirs(OUTPUT_DIR, exist_ok=True)
df.to_csv(OUTPUT_FILE, index=False)
print(f"Cleaned data exported to: {OUTPUT_FILE}")
print(f"File size: {os.path.getsize(OUTPUT_FILE) / 1024:.1f} KB")
```

- **Missing values:** Đã impute 6 cột categorical/text với `"Unknown"` (Chronic_Conditions, Drug_Allergies, Genetic_Disorders, Recommended_Medication, Dosage, Duration).
- **Duplicates:** 0 - không có hàng trùng lặp.
- **Outliers:** Không có giá trị vô lý ngoài khoảng sinh học.
- **Labels:** Cột `Diagnosis` được giữ nguyên (5 lớp).
