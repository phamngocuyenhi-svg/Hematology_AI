# Clean — Rheumatic & Autoimmune Disease Dataset

**Vấn đề từ EDA**: missing 9–43% ở các cột huyết thanh học (ESR, CRP, RF, Anti-CCP...).

**Quyết định**: **giữ nguyên NaN** (không impute — thiếu do không chỉ định xét nghiệm, impute sẽ tạo giá trị giả). Thêm cột flag `<col>_missing` cho các cột quan trọng (CRP, ESR) để model có thể tận dụng thông tin "có đo hay không".

**Kiểm tra bổ sung**: xác nhận không có lỗi encoding ở nhãn Disease (Sjögren's Syndrome) — đã verify ký tự 'ö' đúng Unicode U+00F6, chỉ là hiển thị sai trên console, dữ liệu gốc không lỗi.

**Output**: `Clean_Data/tabular/rheumatic_dataverse_clean.csv`

```python
import pandas as pd
from pathlib import Path

DATA = Path(r'D:\AI_08_V1\Data')
OUT = Path(r'D:\AI_08_V1\Clean_Data\tabular')

df = pd.read_excel(DATA / 'tabular' / 'rheumatic_dataverse' / 'Rheumatic and Autoimmune Disease Dataset.xlsx')
print('Shape gốc:', df.shape)
print('Dòng trùng lặp:', df.duplicated().sum())

df_clean = df.copy()
serology_cols = ['ESR', 'CRP', 'RF', 'Anti-CCP', 'HLA-B27', 'ANA', 'Anti-Ro', 'Anti-La', 'Anti-dsDNA', 'Anti-Sm', 'C3', 'C4']
for col in serology_cols:
    df_clean[f'{col}_missing'] = df_clean[col].isna().astype(int)

print('\nCác cột flag missing mới:', [c for c in df_clean.columns if c.endswith('_missing')])
df_clean.to_csv(OUT / 'rheumatic_dataverse_clean.csv', index=False)
print(f'\nĐã lưu: {OUT / "rheumatic_dataverse_clean.csv"} ({len(df_clean)} dòng, {df_clean.shape[1]} cột)')
```
