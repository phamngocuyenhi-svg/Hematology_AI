# Clean — Anemia Dataset

**Vấn đề từ EDA**: 62.4% dòng trùng lặp hoàn toàn (887/1.421).

**Quyết định**: `drop_duplicates()`, tính lại phân bố nhãn trên dữ liệu sạch.

**Output**: `Clean_Data/tabular/anemia_dataset_clean.csv`

```python
import pandas as pd
from pathlib import Path

DATA = Path(r'D:\AI_08_V1\Data')
OUT = Path(r'D:\AI_08_V1\Clean_Data\tabular')

df = pd.read_csv(DATA / 'tabular' / 'anemia_ml_basic_repo' / 'anemia data from Kaggle.csv')
print('Shape gốc:', df.shape)

df_clean = df.drop_duplicates().reset_index(drop=True)
print(f'Sau loại trùng: {len(df_clean)} dòng ({len(df_clean)/len(df)*100:.1f}% giữ lại)')

print('\nPhân bố nhãn Result — TRƯỚC (có trùng lặp):')
print(df['Result'].value_counts(normalize=True).round(3))
print('\nPhân bố nhãn Result — SAU khi loại trùng:')
print(df_clean['Result'].value_counts(normalize=True).round(3))

df_clean.to_csv(OUT / 'anemia_dataset_clean.csv', index=False)
print(f'\nĐã lưu: {OUT / "anemia_dataset_clean.csv"} ({len(df_clean)} dòng)')
```
