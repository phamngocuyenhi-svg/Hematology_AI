# Clean — Kaggle COVID-19 CBC

**Vấn đề từ EDA**: dataset đã sạch (không missing, không trùng lặp), chỉ có tên cột chưa chuẩn (khoảng trắng thừa, tên dài).

**Quyết định**: chuẩn hóa tên cột, không cần xử lý gì thêm.

**Output**: `Clean_Data/tabular/kaggle_covid_cbc_clean.csv`

```python
import pandas as pd
from pathlib import Path

DATA = Path(r'D:\AI_08_V1\Data')
OUT = Path(r'D:\AI_08_V1\Clean_Data\tabular')

df = pd.read_csv(DATA / 'tabular' / 'kaggle_covid_cbc' / 'COVID-19_CBC_Data.csv')
print('Cột gốc:', list(df.columns))

rename_map = {
    'Admission_DATE ': 'admission_date', 'Discharge_DATE or date of Death': 'discharge_or_death_date',
    'Outcome': 'outcome', 'Patient Age': 'age', 'Gender': 'gender',
    'Sample Collection Date ': 'sample_collection_date',
    'What kind of Treatment provided ': 'treatment', 'Ventilated (Y/N)': 'ventilated',
    'Red blood cell distribution width': 'rdw', 'Monocytes(%)': 'monocytes_pct',
    'White blood cell count': 'wbc', 'Platelet Count': 'platelet_count',
    'Lymphocyte Count': 'lymphocyte_count', 'Neutrophils Count': 'neutrophils_count',
}
df_clean = df.rename(columns=rename_map)
print('\nCột sau chuẩn hóa:', list(df_clean.columns))

df_clean.to_csv(OUT / 'kaggle_covid_cbc_clean.csv', index=False)
print(f'\nĐã lưu: {OUT / "kaggle_covid_cbc_clean.csv"} ({len(df_clean)} dòng, không thay đổi số dòng)')
```
