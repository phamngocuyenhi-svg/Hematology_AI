# Clean — PBC Barcelona

**Vấn đề từ EDA**: đã cân bằng tốt (~2.7 lần), đã xoá 1 file rác `.DS_*` + thư mục `__MACOSX/` thủ công trước đó.

**Quyết định**: quét toàn vẹn 17.093 ảnh bằng `PIL.Image.verify()` để chắc chắn không còn file hỏng nào khác, tạo manifest sạch.

**Output**: `Clean_Data/tabular/pbc_barcelona_manifest.csv`

```python
import pandas as pd
from pathlib import Path
from PIL import Image

ROOT = Path(r'D:\AI_08_V1\Data\images\pbc_barcelona\PBC_dataset_normal_DIB')
OUT = Path(r'D:\AI_08_V1\Clean_Data\tabular')

records = []
corrupt = []
for cls_dir in sorted(ROOT.iterdir()):
    if not cls_dir.is_dir():
        continue
    for f in cls_dir.glob('*.jpg'):
        if f.name.startswith('.'):
            continue
        try:
            with Image.open(f) as im:
                im.verify()
            records.append({'image_path': str(f), 'clean_class': cls_dir.name})
        except Exception as e:
            corrupt.append((str(f), str(e)))

manifest = pd.DataFrame(records)
print(f'Tổng ảnh hợp lệ: {len(manifest)}')
print(f'Số ảnh hỏng phát hiện: {len(corrupt)}')
for c in corrupt:
    print(' ', c)

manifest.to_csv(OUT / 'pbc_barcelona_manifest.csv', index=False)
print(f'\nĐã lưu: {OUT / "pbc_barcelona_manifest.csv"} ({len(manifest)} dòng)')
```
